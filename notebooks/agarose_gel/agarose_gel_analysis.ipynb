```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))  # Set the figure size to make the plot bigger

x = range(1, 8)
y = [(xi - 1) * 1.3 / (((xi - 1) * 1.3) + 7) for xi in x]

plt.bar(x, y)
plt.xlabel('number of secondary fragments', fontsize=22)
plt.ylabel('relative band intensity', fontsize=22)
#plt.title('Insertion step & relative band intensity', fontsize=18)

# Add labels to the bars
for i, value in enumerate(y):
    plt.text(x[i], value + 0.01, f'{value:.2f}', ha='center', fontsize=20)

plt.xticks(fontsize=20)
plt.yticks(fontsize=20)

# Remove the top and right spines
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.show()

# Save the plot as a high-resolution image
#plt.savefig('high_resolution_plot.png', dpi=300)
```
