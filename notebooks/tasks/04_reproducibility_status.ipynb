# Repository reproducibility matrix

Reports maintained, GUI, archive, blocked, deferred, and failed workflows without fabricating missing historical inputs.

**Rules:** `legacy-optimized-v1`; **seed:** 42 unless explicitly noted.

```python
REPO = '/home/wendai/projects/hurdler/clone_repeat_protein'
RULE_PROFILE = 'legacy-optimized-v1'
STATUS_TABLE = '/home/wendai/projects/hurdler/clone_repeat_protein/studies/hurdler_validation/step05_reproducibility/tables/execution_status.csv'
```

```python
from pathlib import Path
import hashlib, json
import pandas as pd

def sha256(path):
    path = Path(path)
    if not path.is_file(): return None
    h = hashlib.sha256()
    with path.open('rb') as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b''): h.update(chunk)
    return h.hexdigest()

run_context = {'rule_profile': RULE_PROFILE, 'input_hashes': {}, 'row_counts': {}, 'filter_flow': [], 'limitations': []}
```

```python
status = pd.read_csv(STATUS_TABLE)
run_context['input_hashes']['execution_status.csv'] = sha256(STATUS_TABLE)
run_context['row_counts']['workflow_rows'] = len(status)
run_context['filter_flow'] = ['inventory canonical, maintained, adjacent, and archive notebook/script entry points', 'retain exact pass/block/defer/fail evidence']
run_context['limitations'] = ['blocked_missing_input results are never imputed']
status.groupby(['status','workflow_type']).size().rename('count').reset_index()
```
