```python
from Bio.Data import CodonTable
import random

def random_reverse_translate(protein_seq, table_id=1):
    """
    Reverse translate a protein sequence into a DNA sequence using random codons.

    Args:
        protein_seq (str): Protein sequence using one-letter amino acid codes.
        table_id (int): NCBI translation table ID (default is 1, the Standard Code).

    Returns:
        str: DNA sequence (codons) corresponding to the protein sequence.
    """
    standard_table = CodonTable.unambiguous_dna_by_id[table_id]
    aa_to_codons = {}

    # Build reverse mapping from amino acids to codons
    for codon, aa in standard_table.forward_table.items():
        if aa not in aa_to_codons:
            aa_to_codons[aa] = []
        aa_to_codons[aa].append(codon)

    # Handle stop codons
    stop_codons = standard_table.stop_codons
    aa_to_codons['*'] = stop_codons

    dna_seq = ''
    for aa in protein_seq:
        aa = aa.upper()
        if aa not in aa_to_codons:
            raise ValueError(f"Invalid or unsupported amino acid: {aa}")
        codon = random.choice(aa_to_codons[aa])
        dna_seq += codon

    return dna_seq
```

```python
def check_13mer(seq):

    oligo_list = []
    for i in range(len(seq)-12):
        oligo_list.append(seq[i:i+13])
    total_oligo_num = len(oligo_list)

    oligo_list = list(set(oligo_list))
    unique_oligo_num = len(oligo_list)
    
    if total_oligo_num == unique_oligo_num:
        print('All 13mers are unique')
    else:
        print('There are some 13mers that are not unique')
        print('Total number of 13mers:', total_oligo_num)
        print('Number of unique 13mers:', unique_oligo_num)
        print('Number of non-unique 13mers:', total_oligo_num - unique_oligo_num)
        print('Non-unique 13mers:', [x for x in oligo_list if oligo_list.count(x) > 1])
```
