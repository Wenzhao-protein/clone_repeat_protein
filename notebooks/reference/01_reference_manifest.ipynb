# Reference manifest

Hashes the maintained reference database. Database generation is kept separate from HURDLER task calls.

**Rules:** `legacy-optimized-v1`; **seed:** 42 unless explicitly noted.

```python
REPO = '/home/wendai/projects/hurdler/clone_repeat_protein'
RULE_PROFILE = 'legacy-optimized-v1'
REFERENCE_DIR = '/home/wendai/projects/hurdler/clone_repeat_protein/data/reference_output'
OUTPUT = '/home/wendai/projects/hurdler/clone_repeat_protein/studies/hurdler_validation/step01_reference_lookup/tables/reference_manifest.json'
```

```python
from pathlib import Path
import hashlib, json
import pandas as pd

def sha256(path):
    path = Path(path)
    if not path.is_file(): return None
    h = hashlib.sha256()
    with path.open('rb') as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b''): h.update(chunk)
    return h.hexdigest()

run_context = {'rule_profile': RULE_PROFILE, 'input_hashes': {}, 'row_counts': {}, 'filter_flow': [], 'limitations': []}
```

```python
from hurdler.reference import build_reference_manifest
manifest = build_reference_manifest(REFERENCE_DIR, OUTPUT)
run_context['input_hashes'] = {row['name']: row['sha256'] for row in manifest['files']}
run_context['row_counts']['reference_files'] = len(manifest['files'])
run_context['filter_flow'] = ['hash every maintained reference file; no implicit download or replacement']
pd.DataFrame(manifest['files'])
```

The manifest records exact content hashes and row counts. It does not download or silently replace reference files.
