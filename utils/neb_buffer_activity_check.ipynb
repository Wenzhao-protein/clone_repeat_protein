```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('./input/neb_buffer_activity.csv')
df['notes'] = df['notes'].astype(str)
df['notes'] = df['notes'].apply(lambda x: x.split(', '))
```

```python
df['ligation_efficiencies'] = df['notes'].apply(lambda x: 'nan' if 'e' in x else 
                                                ('low' if 'a' in x else 
                                                 ('intermediate' if 'b' in x else 'high')))
df['recutting_efficiencies'] = df['notes'].apply(lambda x: 'nan' if 'e' in x else 
                                                ('low' if 'c' in x else 
                                                 ('intermediate' if 'd' in x else 'high')))
df['star_activity'] = df['notes'].apply(lambda x: True if any(char in x for char in ['1', '2', '3']) else False)
df['star_activity_r1.1'] = df['r1.1'].apply(lambda x: True if x[-1]=="*" else False)
df['star_activity_r2.1'] = df['r2.1'].apply(lambda x: True if x[-1]=="*" else False)
df['star_activity_r3.1'] = df['r3.1'].apply(lambda x: True if x[-1]=="*" else False)
df['star_activity_rCutSmart'] = df['rCutSmart'].apply(lambda x: True if x[-1]=="*" else False)
```

```python
def clean_numbers(x):
    
    if isinstance(x, str):
        x = x.rstrip('*+°C')
        x = ''.join(filter(str.isdigit, x))
    if x == '' or x == '<10' or int(x) < 10:
        return 0
    return int(x)
```

```python
df['r1.1'] = df['r1.1'].apply(clean_numbers)
df['r2.1'] = df['r2.1'].apply(clean_numbers)
df['r3.1'] = df['r3.1'].apply(clean_numbers)
df['rCutSmart'] = df['rCutSmart'].apply(clean_numbers)
df['heat_inac'] = df['heat_inac'].apply(clean_numbers)
df['incu_temp'] = df['incu_temp'].apply(clean_numbers)

df.drop(columns=['notes'], inplace=True)
```

```python
df.to_csv('./output/neb_buffer_activity_cleaned.csv', index=False)
df = pd.read_csv('./output/neb_buffer_activity_cleaned.csv')
df
```
